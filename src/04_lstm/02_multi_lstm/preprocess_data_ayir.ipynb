Veri Setini Sektörel Olarak Ayır

```python
import pandas as pd
import os

DATASET_PATH = "../../../data/raw/company_esg_financial_dataset.csv"

# =========================
# 1. Veri yükleme
# =========================
df = pd.read_csv(DATASET_PATH)

print(f"Orijinal veri: {len(df):,} satır")

# =========================
# 2. Sadece GrowthRate eksik doldur
# =========================
df['GrowthRate'] = df.groupby('CompanyID')['GrowthRate'].transform(
    lambda x: x.fillna(x.median())
)

print(f"GrowthRate dolduruldu → kalan NaN: {df['GrowthRate'].isna().sum()}")

# =========================
# 3. Hedef klasör 
# =========================
output_folder = r"../../../data/by_sector"

# =========================
# 4. Sektörlere ayır ve kaydet
# =========================
for sector in df['Industry'].unique():
    sector_df = df[df['Industry'] == sector].copy()
    
    safe_name = sector.replace(" ", "_")
    filename = f"Sector_{safe_name}.csv"
    full_path = os.path.join(output_folder, filename)
    
    sector_df.to_csv(full_path, index=False)
    
    company_count = sector_df['CompanyID'].nunique()
    print(f"{sector:18} → {len(sector_df):5} satır ({company_count} şirket × 11 yıl) → {filename}")

print("\n" + "="*70)
print("TÜM İŞLEMLER TAMAMLANDI!")
print(f"Toplam 9 sektör → {output_folder}")
print("="*70)
```
